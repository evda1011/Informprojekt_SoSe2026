# **Markt-Analyse**
## Konsumverbrauch von Zigaretten (1965–2025)

### **Gesamtziel:**     
Zusammenhang zwischen Tabaksteuern, Preisen & Zigarettenkonsum in Deutschland (1975–2025) untersuchen.


### **Frage 1 (Trend & Faktoren)**    
Wie hat sich der Konsum von Zigaretten pro Einwohner zwischen 1965 und 2025 verändert und welche Faktoren könnten die Trendwende erklären?

### **Fazit:**   
Der Konsum sank von ca. 2.100 Zigaretten pro Kopf (1970) auf 795 (2025) – ein Rückgang um über 60 %. Hauptfaktoren sind steigende Preise (Steuererhöhungen), Werbe- und Rauchverbote sowie die Verlagerung auf E-Zigaretten.

Der langfristige Rückgang ist eindeutig, aber seit 2023 flacht die Kurve ab – der Konsum stagniert auf historischem Tief.

### **Frage 2 (Steuer-Wirksamkeit):**    
Wie wirksam ist die Tabaksteuer als Steuerungsinstrument im deutschen Zigarettenmarkt?

### **Fazit:**
Die Korrelation zwischen Preis und Konsum ist mit **r = -0,94** extrem stark – Preiserhöhungen senken den Konsum. Der Steueranteil liegt seit 1990 konstant bei ca. 58–60 %, doch seit 2023 zeigt der Preismechanismus kaum noch Wirkung.

Die Tabaksteuer ist ein sehr wirksames, aber nicht mehr ausreichendes Instrument. Am „harten Kern“ der Raucher versagt der Preiseffekt – ergänzende Maßnahmen (Aromenverbot, Werbeverbote) werden zunehmend entscheidend
