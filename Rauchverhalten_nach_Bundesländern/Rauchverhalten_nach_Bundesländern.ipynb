## Rauchverhalten nach Bundesländern
## Forschungsfrage

Wie unterscheidet sich der Raucheranteil zwischen den deutschen Bundesländern?

## Ziel

Ziel dieser Analyse ist es, Unterschiede im Raucheranteil zwischen den Bundesländern Deutschlands zu untersuchen.

```python
max_wert = df["Wert"].max()

min_wert = df["Wert"].min()

bundesland_max = df.loc[
    df["Wert"].idxmax(),
    "Region"
]

bundesland_min = df.loc[
    df["Wert"].idxmin(),
    "Region"
]

print(
    f"Den höchsten Raucheranteil hat {bundesland_max}."
)

print(
    f"Den niedrigsten Raucheranteil hat {bundesland_min}."
)
```

```python
if max_wert - min_wert > 15:

    print(
        "Die Unterschiede zwischen den Bundesländern sind deutlich."
    )

else:

    print(
        "Die Unterschiede zwischen den Bundesländern sind gering."
    )
```

```python
import matplotlib.pyplot as plt

farben = plt.cm.Pastel1.colors

plt.figure(figsize=(10, 10))

plt.pie(
    df["Wert"],
    labels=df["Region"],
    autopct="%1.1f%%",
    colors=farben,
    startangle=90
)

plt.title(
    "Raucheranteil nach Bundesland"
)

plt.tight_layout()

plt.show()
```

## Fazit

Die Analyse zeigt, dass das Rauchverhalten in Deutschland regional unterschiedlich ist. Mecklenburg-Vorpommern hat den höchsten Raucheranteil, während das Saarland den niedrigsten Wert aufweist.
