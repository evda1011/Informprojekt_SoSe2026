```python
## Dateipfade festlegen
folder_path = Path.cwd() / "__Prognose_der_Zigarrttenpreis__"
preis_datei = folder_path / "_durchschnittlicher-zigarettenpreis-in-deutschland.xlsx"
konsum_datei = folder_path / "_pro-kopf-verbrauch-von-zigaretten-in-deutschland.xlsx"

## Tabellen einlesen und vorbereiten
def create_table(datei_path, spaltenname):
    df = pd.read_excel(datei_path, sheet_name="Daten", header=None)
    df = df[[1, 2]].dropna()
    df.columns = ["year", spaltenname]
    return df
```
