## Forschungsfrage

Wie unterscheidet sich der Tabakkonsum in Deutschland nach Altersgruppen und Geschlecht?

## Ziel

Ziel dieser Analyse ist es, Unterschiede im Tabakkonsum zwischen verschiedenen Altersgruppen und zwischen Männern und Frauen zu untersuchen.

```python
import pandas as pd

from src.plot_utils import plot_balken_gruppiert
```

```python
## Daten einlesen
df = pd.read_excel(
    "statistic_id596838_tabakraucher-in-deutschland-nach-altersgruppen-bis-2024_daten bereinigung.xlsx"
)

df
```

```python
## Datenanalyse
letztes_jahr = df.iloc[-1]

print("Werte für 2024:")

print(
    f"14–17 Jahre: {letztes_jahr['14_17']} %"
)

print(
    f"18–24 Jahre: {letztes_jahr['18_24']} %"
)

print(
    f"25+ Jahre: {letztes_jahr['25_plus']} %"
)





if letztes_jahr["18_24"] > letztes_jahr["25_plus"]:

    print(
        "Die Altersgruppe 18–24 Jahre weist den höchsten Tabakkonsum auf."
    )

else:

    print(
        "Die Altersgruppe ab 25 Jahren weist den höchsten Tabakkonsum auf."
    )
```

```python
## Visualisierung
plot_balken_gruppiert(

    df,

    "Jahr",

    [

        "14_17",

        "18_24",

        "25_plus"

    ],

    "Tabakkonsum nach Altersgruppen",

    "Jahr",

    "Anteil [%]"
)
```

## Interpretation

Das Diagramm zeigt die Entwicklung des Tabakkonsums in verschiedenen Altersgruppen von 2016 bis 2024.
