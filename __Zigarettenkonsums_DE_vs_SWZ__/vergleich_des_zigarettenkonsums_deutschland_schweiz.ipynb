## Vergleich des zigarettenkonsums Deutschland und Schweiz

### **Frage:**   
Wie unterscheidet sich der Zigarettenkonsum zwischen Deutschland und der Schweiz in verschiedenen Altersgruppen?

### **Ziel der Analyse:**    
Ziel dieser Analyse ist es, den Zigarettenkonsum in Deutschland und der Schweiz zu vergleichen. Dabei soll untersucht werden, in welchen Altersgruppen Unterschiede bestehen und welches Land insgesamt einen höheren Konsum aufweist.


```python
# imports
import sys
from pathlib import Path
import pandas as pd

projekt_ordner = Path.cwd().parent

sys.path.append(str(projekt_ordner))

from src.plot_utils import (
    plot_balken_gruppiert,
    plot_kreis)
```

```python
## Daten einlesen
deu_datei = "DEU.xlsx"
schw_datei = "SCHW.xlsx"

deu = pd.read_excel(deu_datei)
schw = pd.read_excel(schw_datei)


comparison = pd.DataFrame({
    "Altersgruppe": deu["Altersgruppe"],
    "Deutschland": deu["DE_Insgesamt"],
    "Schweiz": schw["CH_Insgesamt"]
})

comparison
```

### **Datenanalyse**    
Die Daten werden mit einer Schleife analysiert. Dabei wird für jede Altersgruppe überprüft, welches Land einen höheren Zigarettenkonsum aufweist.

```python
for index, row in comparison.iterrows():

    if row["Deutschland"] > row["Schweiz"]:
        print( f"{row['Altersgruppe']}: "
            "Deutschland hat höheren Konsum.")

    else:
        print(f"{row['Altersgruppe']}: "
            "Schweiz hat höheren Konsum.")
```

```python
## Balkendiagramm
plot_balken_gruppiert(
    comparison,
    "Altersgruppe",
    ["Deutschland", "Schweiz"],
    "Zigarettenkonsum nach Altersgruppen: Deutschland vs. Schweiz",
    "Altersgruppe",
    "Anteil [%]"
)
```

```python
## Kreisdiagramm
de_mean = comparison["Deutschland"].mean()

ch_mean = comparison["Schweiz"].mean()

labels = ["Deutschland", "Schweiz"]

werte = [de_mean, ch_mean]

farben = ["#0E6A64","#5DC6D0"]

plot_kreis( labels, werte, "Vergleich Deutschland vs. Schweiz insgesamt", farben)
```

### **Fazit:**  
Die Analyse zeigt, dass sich der Zigarettenkonsum zwischen Deutschland und der Schweiz je nach Altersgruppe unterscheidet. Durch die Berechnung der Durchschnittswerte und die grafische Darstellung können diese Unterschiede leicht erkannt werden.
